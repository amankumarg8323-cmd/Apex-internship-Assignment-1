```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Reproducibility
np.random.seed(42)
```

## 1. NumPy Fundamentals

```python
# Create 1D and 2D arrays
array_1d = np.array([10, 20, 30, 40, 50])
array_2d = np.array([[1, 2, 3], [4, 5, 6]])

print("1D array:", array_1d)
print("2D array:\n", array_2d)

# Generate arrays using zeros(), ones(), arange(), and linspace()
print("Zeros:\n", np.zeros((2, 3)))
print("Ones:", np.ones(4))
print("Arange:", np.arange(0, 10, 2))
print("Linspace:", np.linspace(0, 1, 5))

# Indexing and slicing
print("First element:", array_1d[0])
print("Elements from index 1 to 3:", array_1d[1:4])
print("2D array, first row:", array_2d[0, :])

# Mathematical operations
print("Add 5:", array_1d + 5)
print("Multiply by 2:", array_1d * 2)
print("Square:", array_1d ** 2)

# Statistical calculations
print("Mean:", np.mean(array_1d))
print("Median:", np.median(array_1d))
print("Standard deviation:", np.std(array_1d))
print("Minimum:", np.min(array_1d))
print("Maximum:", np.max(array_1d))

# Reshape and flatten
reshaped = np.arange(1, 7).reshape(2, 3)
print("Reshaped array:\n", reshaped)
print("Flattened array:", reshaped.flatten())
```

## 2. Create and Explore a Retail Dataset

```python
# Synthetic retail data for demonstration only
data = {
    "Order_ID": range(1001, 1013),
    "Branch": ["North", "South", "East", "West"] * 3,
    "Product": ["Laptop", "Phone", "Tablet", "Headphones"] * 3,
    "Category": ["Electronics", "Electronics", "Electronics", "Accessories"] * 3,
    "Quantity": [2, 3, 1, 5, 1, 2, 4, 3, 2, 1, 3, 2],
    "Unit_Price": [55000, 18000, 25000, 2000, 52000, 19000, 24000, 2200, 56000, 17500, 26000, 2100],
    "Sales": [110000, 54000, 25000, 10000, 52000, 38000, 96000, 6600, 112000, 17500, 78000, 4200],
    "Customer_Type": ["New", "Returning", "New", "Returning"] * 3
}
df = pd.DataFrame(data)

# Save a CSV copy
df.to_csv("retail_sales_sample.csv", index=False)

# Import a CSV dataset
df = pd.read_csv("retail_sales_sample.csv")

print("First 10 records:\n", df.head(10))
print("Last 10 records:\n", df.tail(10))
print("Dataset shape:", df.shape)
print("Column names:", df.columns.tolist())
print("Data types:\n", df.dtypes)
print("Summary statistics:\n", df.describe())
print("Dataset information:")
df.info()
```

## 3. Data Operations

```python
# Select specific columns
print(df[["Product", "Quantity", "Sales"]])

# Create a calculated column: average revenue per item
df["Revenue_Per_Item"] = df["Sales"] / df["Quantity"]

# Filter rows based on a condition
high_sales = df[df["Sales"] > 50000]
print("Orders with sales above 50,000:\n", high_sales)

# Count unique values
print("Unique products:", df["Product"].nunique())
print("Unique branches:", df["Branch"].nunique())

# Sort dataset
sorted_df = df.sort_values(by="Sales", ascending=False)
print("Sorted by sales:\n", sorted_df)

# Rename columns
renamed_df = df.rename(columns={"Sales": "Total_Sales", "Branch": "Branch_Name"})
print("Renamed columns:", renamed_df.columns.tolist())

# Find missing values
print("Missing values per column:\n", df.isnull().sum())
```

## 4. Basic Data Analysis

```python
total_records = len(df)
total_columns = len(df.columns)

print("Total records:", total_records)
print("Total columns:", total_columns)
print("Average sales:", df["Sales"].mean())
print("Highest sales:", df["Sales"].max())
print("Lowest sales:", df["Sales"].min())
print("Customer type frequency:\n", df["Customer_Type"].value_counts())
print("Top 5 orders by sales:\n", df.nlargest(5, "Sales"))
```

## 5. Data Visualization

```python
# Bar chart: total sales by product
product_sales = df.groupby("Product")["Sales"].sum()
product_sales.plot(kind="bar", title="Total Sales by Product", xlabel="Product", ylabel="Sales")
plt.tight_layout()
plt.show()

# Line chart: sales by order
plt.plot(df["Order_ID"], df["Sales"], marker="o")
plt.title("Sales by Order")
plt.xlabel("Order ID")
plt.ylabel("Sales")
plt.grid(True)
plt.tight_layout()
plt.show()

# Histogram: distribution of sales
plt.hist(df["Sales"], bins=6, edgecolor="black")
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# Pie chart: customer type share
df["Customer_Type"].value_counts().plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("Customer Type Distribution")
plt.ylabel("")
plt.tight_layout()
plt.show()

# Box plot: sales spread
plt.boxplot(df["Sales"])
plt.title("Box Plot of Sales")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

## 6. Business Insights

### Five observations
1. The sample dataset contains 12 retail orders across four branches.
2. The dataset contains four products: Laptop, Phone, Tablet and Headphones.
3. Laptop orders contribute the largest sales amounts in this sample.
4. Sales values vary considerably between orders.
5. Both New and Returning customers are represented.

### Three data challenges
1. The dataset is small and may not represent actual customer behaviour.
2. It covers only a limited number of branches, products and orders.
3. It has no date, customer identifier, or customer demographic fields, limiting trend and customer-level analysis.

### Three recommendations for data quality
1. Collect more records across more dates, branches and product categories.
2. Validate quantity, unit price and sales values, and ensure sales equals quantity multiplied by unit price.
3. Add consistent order dates and anonymized customer IDs; check missing values and duplicate order IDs before analysis.
